## 1. group by

```python
import pandas as pd
```

```python
insurance_details = pd.read_csv("insurance.csv")
insurance_details
```

### averg ins charge region vise

```python
insurance_details.groupby(by = "region")["charges"].mean().round().sort_values()
```

```python
insurance_details.groupby(by = ["region","sex"])["charges"].mean().round().sort_values()
```

## 2. Pivot Table
Pivot tables in pandas let you summarize, group, and reshape your DataFrame—similar to Excel pivot tables but fully programmable. They are ideal when you want categories on rows and columns with aggregated values in the cells.

```python
pd.pivot_table(data = insurance_details,values="charges",index = "region", columns = ["sex",'children'])
```

## 3. Aggregrate

```python
insurance_details.groupby(by = ["region","sex"])["charges"].aggregate(func=["mean","min","max","sum"]).round()
```

## 4. CrossTab
pandas.crosstab() creates a frequency table (contingency table) that summarizes the relationship between two or more categorical variables. It is ideal when you want quick counts or normalized percentages across categories

```python
pd.crosstab(index = insurance_details["region"],columns=insurance_details["smoker"],margins=True)
```

## 6. Apply

#### Create a new column eligibility criteria where age >20 is eligible else not not eligible

```python
def check_age(age):
    if age > 20:
        return "eligible"
    else:
        return "Not eligible"
```

```python
insurance_details["eligibility criteria"] = insurance_details["age"].apply(check_age)
```

```python
insurance_details
```

## 7. Lambda

```python
insurance_details["ec2"] = insurance_details["age"].apply(lambda x:"Eligible" if x>20 else "not Eligible")
```
