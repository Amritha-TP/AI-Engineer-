```python
import pandas as pd

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from sklearn.svm import SVC

from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
```

```python
cancer_data = load_breast_cancer()
cancer_df = pd.DataFrame(cancer_data.data,columns=cancer_data.feature_names)
cancer_df["target"]=cancer_data.target
cancer_df
```

```python
X=cancer_df.drop("target",axis=1)
y=cancer_df["target"]
```

```python
std_scalar = StandardScaler()
scalar_X=std_scalar.fit_transform(X)
pd.DataFrame(scalar_X)
```

```python
X_train, X_test, y_train, y_test=train_test_split(scalar_X,y,test_size=0.2,shuffle=True,stratify=y,random_state=42)
```

```python
vec_classifier=SVC(kernel='rbf')
vec_classifier.fit(X_train,y_train)
```

```python
y_pred_train = vec_classifier.predict(X_train)
```

```python
print(accuracy_score(y_train,y_pred_train))
```

```python
print(classification_report(y_train,y_pred_train))
```

```python
y_pred_test = vec_classifier.predict(X_test)
```

```python
print(accuracy_score(y_test,y_pred_test))
```

```python
print(classification_report(y_test,y_pred_test))
```
