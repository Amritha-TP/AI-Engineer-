```python
import pandas as pd #To read files
from matplotlib import pyplot as plt # Used for graph data visualization 
import seaborn as sns # Data Visualization Library
from sklearn.linear_model import LinearRegression # supports LinearRegression.You can then create a model, fit it on data, and make predictions.
import statsmodels.api as sm
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error,mean_squared_error,root_mean_squared_error,r2_score # Linear regression evaluation metrics
import statsmodels.formula.api as smf

import warnings
warnings.filterwarnings("ignore") # to ignore warnings
```

```python
cars_data = pd.read_csv(filepath_or_buffer = "Cars.csv")
cars_data
```

```python
cars_data.isna().sum()
```

```python
cars_data.dtypes
```

```python
X = cars_data[["HP","VOL","SP","WT"]]
X
```

```python
y = cars_data["MPG"]
y
```

```python
linear_model = LinearRegression()
```

```python
X_train, X_test, y_train, y_test=train_test_split(X,y,test_size=0.20,shuffle=True,random_state=34)
```

```python
linear_model.fit(X_train,y_train)
```

```python
y_pred_train = linear_model.predict(X_train)
y_pred_train
```

```python
y_pred_test = linear_model.predict(X_test)
y_pred_test
```

## train evl metrics

```python
mean_absolute_error(y_train,y_pred_train)
```

```python
mean_squared_error(y_train,y_pred_train)
```

```python
root_mean_squared_error(y_train,y_pred_train)
```

```python
r2_score(y_train,y_pred_train)
```

## test evl metrics

```python
mean_absolute_error(y_test,y_pred_test)
```

```python
mean_squared_error(y_test,y_pred_test)
```

```python
root_mean_squared_error(y_test,y_pred_test)
```

```python
r2_score(y_test,y_pred_test)
```

```python
linear_stats_model = smf.ols(formula = "MPG ~ HP+VOL+SP+WT", data=cars_data).fit()
```

```python
print("r2 score:",linear_stats_model.rsquared.round(2),"\nAdjusted r2: ",linear_stats_model.rsquared_adj.round(2))
```

```python
train_data = X_train.join(y_train)
train_data
```

```python
linear_stats_model_train = smf.ols(formula = "MPG ~ HP+VOL+SP+WT", data=train_data).fit()
```

```python
print("r2 score train:",linear_stats_model_train.rsquared.round(2),"\nAdjusted r2 train: ",linear_stats_model_train.rsquared_adj.round(2))
```

```python
test_data = X_test.join(y_test)
test_data
```

```python
test_predictions = linear_stats_model_train.predict(X_test)
```

```python
r2_score(y_test,test_predictions)
```

```python
# 1. Use the true test R2 score you calculated in cell
r2_test = 0.767

# 2. Get n (number of test rows) and k (number of features)
n = X_test.shape[0]  # 17
k = X_test.shape[1] # 4

# 3. Calculate True Adjusted R2
true_adj_r2_test = 1 - ((1 - r2_test) * (n - 1)) / (n - k - 1)

print("True Test Adjusted R2:", round(true_adj_r2_test, 4))
```
