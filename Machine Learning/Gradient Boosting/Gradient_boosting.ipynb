## Data Preparation

```python
X = breast_cancer_df.drop("Target",axis=1)
y = breast_cancer_df["Target"]
```

## Standardizing the input using standard scalar where the mean is 0 and std is 1

```python
std_scalar = StandardScaler()
scalar_X = std_scalar.fit_transform(X)
```

```python
X_train, X_test, y_train, y_test = train_test_split(scalar_X,y,test_size=0.20,random_state=43,shuffle=True,stratify=y)
```

## Model Building

```python
gdc = GradientBoostingClassifier(max_depth=3,learning_rate=0.01)
```

```python
%%time
gdc.fit(X_train,y_train)
```

## Model Testing | Model Evaluation

```python
y_pred_train=gdc.predict(X_train)
```

```python
print(classification_report(y_train,y_pred_train))
```

```python
print(confusion_matrix(y_train,y_pred_train))
```

```python
y_pred_test=gdc.predict(X_test)
```

```python
print(classification_report(y_test,y_pred_test))
```

```python
print(confusion_matrix(y_test,y_pred_test))
```
