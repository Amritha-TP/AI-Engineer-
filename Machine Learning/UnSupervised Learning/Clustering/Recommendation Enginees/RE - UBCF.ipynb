## Data Preparation

#### WHY, WHAT, WHEN

1. Why Put userId as Columns and movie as the Index?
By default, the Pandas .corr() function calculates correlation between columns.

The "What": You pivoted your original long-form dataset into a wide matrix where each column represents a unique user, each row represents a unique movie, and the cells contain the ratings.

The "Why": Since .corr() looks at columns, putting userId in the columns forces Pandas to compare the rating vectors of User A vs. User B across all the movies they have rated.

The "When": You do this specifically when you want to build a User-Based Collaborative Filtering (UBCF) system. It allows you to answer: "Which users have the most similar taste in movies to User X?"

Note: If you wanted to do Item-Based Collaborative Filtering (finding movies similar to other movies), you would do the exact opposite: put movie as columns and userId as the index.

2. Why Check Correlation of userId against userId?
When you run movie_pivot_table.corr(), you generate a User-User Similarity Matrix.

How to Interpret the Matrix:
The Diagonals (1.000000): You will notice a perfect diagonal of 1.0 from the top-left to the bottom-right. This is because every user is perfectly correlated with themselves (e.g., User 3 vs. User 3).

The Off-Diagonals (Values between -1 and 1): These numbers tell you how aligned two different users' tastes are:

Positive correlation (closer to 1): Users rated the same movies similarly. If User A liked it, User B likely liked it too.

Negative correlation (closer to -1): Users have opposite tastes. If User A loved a movie, User B hated it.

Near 0: No apparent relationship between their voting patterns.

Why is this useful for recommendations?
Once you have this userId vs. userId matrix, you can pinpoint the closest "neighbors" for any target user. For example, if you want to recommend a movie to User 3, you look at the column for User 3, find other users with high positive correlation scores (like User 11 or User 6993, which show 1.0 or close to it in your matrix), and recommend highly-rated movies that those similar users watched but User 3 hasn't seen yet.

```python
movie_pivot_table=pd.pivot_table(data=movie_data,columns='userId',index='movie').fillna(0)
movie_pivot_table.columns = movie_data["userId"].unique()
movie_pivot_table
```

## 2. Building UBCF using Euclideon Distance

```python
from sklearn.metrics import pairwise_distances
```

```python
euclideon_piv_table = pd.pivot_table( data= movie_data,values="rating",index="userId", columns="movie",).fillna(0)
euclideon_piv_table.columns = movie_data["movie"].unique()
euclideon_piv_table.index   = movie_data["userId"].unique()
euclideon_piv_table
```

```python
movie_euclidean = pd.DataFrame(pairwise_distances(X = euclideon_piv_table, metric='euclidean'))
movie_euclidean.columns = movie_data["userId"].unique()
movie_euclidean.index   = movie_data["userId"].unique()
movie_euclidean.iloc[:20,:20]
```

```python
movie_data[(movie_data["userId"] == 8) | (movie_data["userId"] == 58)]
```

### 3. Building UBCF using Cosine Similarity

```python
cosine_correlation = pd.DataFrame(1- pairwise_distances(X = euclideon_piv_table, metric='cosine',))
cosine_correlation.columns = movie_data["userId"].unique()
cosine_correlation.index   = movie_data["userId"].unique()
cosine_correlation
```
