```python
del online_retail_data["CustomerID"]
```

```python
online_retail_data.dropna(inplace=True)
```

## Country-wise data

```python
uk_data = online_retail_data[online_retail_data["Country"]=="United Kingdom"]
uk_data
```

```python
uk_data.isna().sum()
```

```python
uk_pivotted_data = pd.pivot_table(data=uk_data,values="Quantity",columns="Description",index="InvoiceNo",aggfunc="sum").fillna(0)
uk_pivotted_data
```

```python
uk_onehot=uk_pivotted_data.applymap(lambda x:0 if x==0 else 1)
uk_onehot
```

```python
uk_onehot.describe()
```

## 4. Data Mining

```python
from mlxtend.frequent_patterns import apriori,association_rules
```

```python
apri_data = apriori(df=uk_onehot,min_support=0.02,use_colnames=True)
apri_data
```

```python
best_associates=association_rules(df=apri_data,metric='lift',min_threshold=0.1)
```

```python
best_associates.to_csv("best_anticidents_consequents.csv")
```
