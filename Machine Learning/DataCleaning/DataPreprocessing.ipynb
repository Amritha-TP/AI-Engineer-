```python
weather_copy03=weather_data.copy()
```

```python
from sklearn.preprocessing import OneHotEncoder
import pandas as pd

# 1. Initialize the encoder (sparse_output=False gives a numpy array)
ohe = OneHotEncoder(sparse_output=False)

# 2. Fit and transform the Weather column
encoded = ohe.fit_transform(weather_copy03[["Weather"]])

# 3. Create the DataFrame, CRUCIALLY passing the original index
encoded_df = pd.DataFrame(
    encoded, 
    columns=ohe.get_feature_names_out(["Weather"]),
    index=weather_copy03.index  # <--- This aligns the rows perfectly!
)

# 4. Concat them side-by-side
weather_copy03 = pd.concat([weather_copy03, encoded_df], axis=1)

# 5. Drop the original string column since it's now encoded
weather_copy03.drop(columns=["Weather"], inplace=True)

weather_copy03.head()
```
