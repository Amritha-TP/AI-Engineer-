```python
import pandas as pd

from sklearn.preprocessing import StandardScaler
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix,classification_report

import xgboost as xgb
```

```python
breast_cancer_datset = load_breast_cancer()
```

```python
breast_cancer_df = pd.DataFrame(data=breast_cancer_datset.data,columns=breast_cancer_datset.feature_names)
```

```python
breast_cancer_df["Target"] = breast_cancer_datset.target
```

```python
breast_cancer_df
```

```python
breast_cancer_df.isna().sum()
```

```python
breast_cancer_df["Target"].value_counts()
```

```python
X= breast_cancer_df.drop("Target",axis=1)
y = breast_cancer_df["Target"]
```

```python
std_scalar=StandardScaler()
scalar_x = std_scalar.fit_transform(X)
scalar_x
```

```python
X_train, X_test, y_train, y_test=train_test_split(scalar_x,y,test_size=0.2,shuffle=True,random_state=43,stratify=y)
```

```python
xg_classifier = xgb.XGBClassifier()
```

```python
xg_classifier.fit(X_train,y_train)
```

```python
y_pred_train=xg_classifier.predict(X_train)
y_pred_train
```

```python
print(classification_report(y_train,y_pred_train))
```

```python
print(confusion_matrix(y_train,y_pred_train))
```

```python
y_pred_test = xg_classifier.predict(X_test)
```

```python
print(classification_report(y_test,y_pred_test))
```

```python
print(confusion_matrix(y_test,y_pred_test))
```
