## HOW TO CREATE CLASS AND OBJECTS?

- Class: A blueprint containing attributes and methods
- Object: Multiple objects can be created to access the class’s methods and variables.
- __init__: Constructor method that runs first when an object is created
- self: Refers to the current object; used to access class variables and methods
- Encapsulation: Achieved using double underscores (__) to hide details
- Abstract Class: Contains methods without definitions; child classes must implement them
- Inheritance: Enables reuse of parent class  use super() to access parent methods/variables
- Polymorphism: Same method name, different behavior {Diff parameters,diff types [assed as argument}
