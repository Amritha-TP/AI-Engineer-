## ObjectsOriented Programming

  OOP is a way of organizing code that uses objects and classes to represent real-world entities and their behavior. 

- Organizes code into classes and objects.
- Supports encapsulation to group data and methods together.
- Enables inheritance for reusability and hierarchy.
- Allows polymorphism for flexible method implementation.
- Improves modularity, scalability and maintainability.

### 1. Inheritance

It extracts methods,properties and attributes from one class (parent class) to other class (Child class).It supports heirarchy and code reusability.

eg: animal (parent class) where dog,cat,racoon are child class of animal

```python
class Animal:
    def __init__(self,name,age):
        self.name = name
        self.age =age
        

    def moves(self):
        print(f"{self.name} is an animal, who navigates by walking on four limbs.")

    def smells(self):
        print(f"{self.name} is an animal, who smells using it's nose.")

class Dog(Animal): # inherits the properties from Animal class
   # def __init__(self,name):
       # super().__init__(name,11)
    def sound(self):
        print(f"{self.name} is a dog and it barks.")

dog1 = Dog("Ranger",12)
dog1.smells() # smells belongs to animal parent class
dog1.sound()# sound belongs to dog child class
dog1.age
```

#### Note: 
- if a child class has no init constructor it looks for constructors in the parent class.
- you initalize parent constructor using super().__init__() inside a child init constructor.

### Types of inheritence

#### 1. Single Inheritance
Single inheritance enables a derived class to inherit properties from a single parent class.

```python
class Parent:
    def method1(self):
        print("I'm Parent class")

class Child(Parent):
    def method2(self):
        print("I'm child class")

c1 = Child()
c1.method1()
c1.method2()
```

#### 2.Multiple Inheritance
- When a class can be derived from more than one parent class.
- all the features of the parent classes are inherited into the child class. 

```python
class Mother:
    mother_name = "aa"

    def mom_method(self):
        print(f"Mom name is {self.mother_name}")

class Father:
    father_name = "dd"

    def dad_method(self):
        print(f"Dad name is {self.father_name}")

class Son(Mother,Father):
    def parents(self):
        print(f"Father name is {self.father_name}")
        print(f"Mother name is {self.mother_name}")

s1 = Son()
s1.father_name="Thiru"
s1.mother_name="seetha"
s1.parents()
s1.mom_method()
s1.dad_method()
```

#### 3.Multilevel Inheritance
- features of the parent class and the child class are further inherited into the new child class.
- similar to a relationship representing a child and a grandfather. 

```python
class Grandfather:
    def __init__(self, grandfathername):
        self.grandfathername = grandfathername


class Father(Grandfather):
    def __init__(self, fathername, grandfathername):
        self.fathername = fathername
        Grandfather.__init__(self, grandfathername)


class Son(Father):
    def __init__(self, sonname, fathername, grandfathername):
        self.sonname = sonname
        Father.__init__(self, fathername, grandfathername)

    def print_name(self):
        print('Grandfather name :', self.grandfathername)
        print('Father name :', self.fathername)
        print('Son name :', self.sonname)

# Driver code
s1 = Son('Ram', 'Thiru', 'Maharaj')
s1.print_name()
```

#### 4. Hierarchical Inheritance
- more than one derived class are created from a single base
- 1parent class - 2 or more child class

```python
class Parent:
    def parent_method(self):
        print("This function is in parent class.")

# Derived class 1
class Child1(Parent):
    def child1_method(self):
        print("This function is in child 1.")

# Derived class 2
class Child2(Parent):
    def child2_method(self):
        print("This function is in child 2.")

# Driver code
object1 = Child1()
object2 = Child2()

object1.parent_method()
object1.child1_method()
object2.parent_method()
object2.child2_method()
```

### 2. Abstraction
-  is used to hide the implementation details from the user and expose only necessary parts.

Abstract Base Class
 - is used to achieve data abstraction by defining a common interface for its subclasses.
 - methods cannot be initiated in abc.
 - forces methods to only be initated in its's child class
 - abc cannot have it's own objects

```python
from abc import ABC,abstractmethod

class Animals1(ABC):
    def __init__(self,name):
        self.name = name

    @abstractmethod
    def sound(self): #abs method without body
        pass

class Dog1(Animals1): # concrete class which is going to implement the abs class method
    def sound(self):
        print(f"{self.name} barks.")

d1 = Dog1("Ranger")

d1.sound()
```

### 3. Polymorphism
- same method name across different classes
- run-time polymorphism works in different classes with same method name (method over riding)
- compile-time works in same class same method name with difference in the number of parameters and data types.(method over loading)

note: Python does NOT support method overloading the way Java or C++ does.
      If two methods have the same name, it takes the last one.

```python
# runtime polymorphism
class GooglePay:
    def pay(self):
        print("paid through google pay.")

class PayPal:
    def pay(self):
        print("paid through paypal.")

gpay = GooglePay()
gpay.pay()
pal = PayPal()
pal.pay()
```

```python
class Add:
    def add1(self,a,b):
        print(a+b)
    def add1(self,a,b,c):
        print(a+b+c)

a = Add()
a.add1(1,2)
a.add1(1,2,3)

#Python does NOT support method overloading the way Java or C++ does.
#If two methods have the same name, it takes the last one.
# it supports only over riding
# *args helps u to acheive compile time polymorphism
```

```python
class Add:
    def add1(self,a,b,*args):
        result = a+b
        for num in args:
            result = result+num
        return result
a = Add()
print(a.add1(1,2))
print(a.add1(1,2,3))
```
