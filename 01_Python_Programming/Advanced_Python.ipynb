### Positional Arguments

```python
def get_details(name, job_title, salary):
    print("Name:",name)
    print("job_title:",job_title)
    print("salary:", salary)
```

```python
get_details("Amritha","AI engineer",4000000)
```

```python
get_details("AI engineer",4000000,"Amritha") # position of arguments matters
```

```python
get_details(job_title="AI engineer",salary=4000000,name="Amritha") # u can itself the align values with respective parameters 
```

### Default arguments

```python
def get_details_aiEgineer(name,salary,job_title="AI Egineer"): # default arguments should be defined in the last
    print("Name:",name)
    print("job_title:",job_title)
    print("salary:", salary)
```

```python
get_details_aiEgineer("Mathew",400000) # no need pass arg for job_title as it is default arg
```

```python
# incase u want to change the default arg, u can change it 
get_details_aiEgineer("Mathew",400000,job_title="ML Egineer") # no need pass arg for job_title as it is default arg
```

### Variable length argument - *parameter_name
- can pass multiple argument ina single paramenter and can't use the default argument when this exists
- should be declared as the last parameter
- only one variable length arg can be declared

```python
def get_details(name, job_title, *salary):
    print("Name:",name)
    print("job_title:",job_title)
    print("salary:", salary)
```

```python
get_details("Krish","ML Engineer",400000,300000,200000)
```

### Keyword argument - **Parameter name

```python
def get_details(name, job_title, **salary):
    print("Name:",name)
    print("job_title:",job_title)
    print("salary:", salary)
```

```python
get_details("Harish","Manufacturing Engineer", sal_1=10000, sal_2=9000, sal_3=8000)
```

## 5.4 challenge to find the factors of a number
the factor of 20 is 1,2,4,5,10,20

```python
user_input = int(input("Enter the number:"))

for i in range(1,user_input+1):
    if user_input%i==0:
        print(i)
```

```python
user_input = int(input("Enter the number:"))
result = set()

for i in range(1,user_input+1):
    if user_input%i==0:
        result.add(i)
print ("The factor of {} is {}".format(user_input,result))
```
