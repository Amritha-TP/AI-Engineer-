## Task 3 Rock Paper scissors


### Objective:
Build a simple Rock-Paper-Scissors game where the player competes with the computer.
 - This task will test your understanding of:
 - Input/output
 - Conditional logic
 - Random module
 - Loops
 - Exception handling

### Instructions:
1. The user should enter their choice: rock, paper, or scissors.
2. The computer should randomly select one of these.
3. Compare both and print who wins.
4. Ask if the player wants to play again- if yes, repeat; if no, exit.
5. Handle invalid inputs gracefully using try-except or condition checks. 

```python
import random

game_props = ["Rock","Paper","Scissors"]


while True:
    try:
        game_bot = random.choice(game_props)
        user_input = input("Enter Rock or Paper or Scissors:").capitalize()

        if (user_input not in game_props):
            print("your input is invalid. Enter a valid input given here",game_props)
            continue
        
        print("Game bot :",game_bot)
        
        if(game_bot==user_input):
            print("It's a tie")
        elif(game_bot == "Rock" and user_input == "Paper"):
            print("Hurray!!, you won the game")
        elif(game_bot == "Rock" and user_input == "Scissors"):
            print("You lost the game :(")
        elif(game_bot == "Paper" and user_input == "Rock"):
                print("You lost the game :(")
        elif(game_bot == "Paper" and user_input == "Scissors"):
                print("Hurray!!, you won the game")
        elif(game_bot == "Scissors" and user_input == "Paper"):
                print("You lost the game :(")
        elif(game_bot == "Scissors" and user_input == "Rock"):
                print("Hurray!!, you won the game")
    except :
        print (" Something went wrong! ")
    finally:
        y_n=input("Do you want to still play?").lower()
        if(y_n =="no"):
            break
        else:
            continue
        
        
    
        
```
