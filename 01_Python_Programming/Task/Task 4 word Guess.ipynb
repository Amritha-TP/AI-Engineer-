## Word Guessing Game (Mini Hangman)
Objective:
Create a word-guessing game where the player tries to guess the letters of a hidden word.

This will test your:
●Loops
●Conditional logic
●String manipulation
●Exception handling
●Logical thinking

Instructions:
1. Create a list of random words (like  “python”, “machine”, “learning”, “bridge”).
2. Randomly select one word for the game.
3. Show the word as blanks (_ _ _ _) and let the player guess one letter at a time.
4. Keep track of wrong guesses (max 6 attempts).
5. End the game when the player either guesses the full word or uses all chances.

## Corrected Approach

```python
import random

words_list = ["python", "machine", "learning","bridge","cat", "swim"]

word = random.choice(words_list)

blanks = ["_"]*len(word)

max_wrong_attempts = 6
wrong_attempts = 0

guessed_letters = []

try:

    while wrong_attempts < max_wrong_attempts:

        try:
            user_input = input("Enter the guess one letter of the word").lower()
        except Exception as e:
            print("Exception Error occured while entering the user_input: ",e)
            continue

        try:
            if len(user_input) !=1 or not user_input.isalpha():
                raise ValueError("Enter only one alphabet. Numbers and other special characters is not allowed")
        except ValueError as v:
            print(v)
            continue
            
    
        if user_input in guessed_letters:
            print (f"you already guessed {user_input}. Enter a non guessed letter")
            continue
            
        guessed_letters.append(user_input)
        
        if user_input in word:
            print ("you guessed right!")
            
            for i in range(0,len(word)):
                if(word[i] == user_input):
                    blanks[i] = word[i]
    
            print(blanks)
    
            if "_" not in blanks:
                break
    
        else:
            wrong_attempts+=1
            print (f"your guess is wrong. you only {max_wrong_attempts-wrong_attempts} worng attempts left")
            
        
    
    if "_" in blanks:
        print(f"wrong word. the right word is {word}")
    else:
        print(f"you guessed the word is right. It's {''.join(blanks)}" )

except Exception as e:
    print("Exception Error: ",e)
          
            
            
```
