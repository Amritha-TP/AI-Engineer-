## 3 Given multiple condition 

1. The user should enter the number between 1-100. otherwise it should say that "The entered number is invalid".
2. If the user guessed number > secert number. --> output: "The number is greater than the secert number".
3. If the user guessed number < secert number. --> output: "The number is lesser than the secert number".

```python
print(" WELCOME TO THE NUMBER GUESSING GAME!! ")
print(" Guess the number between 1 to 100. ")
print("*************************************************")

secert_number = 78

for i in range(0,7):
    user_input = int(input("Enter the number:"))
    if(user_input >100 or user_input < 1):
        print("Your number is invalid enter the number between 1 - 100.")
    elif (user_input > secert_number):
        print("Your number is greater than the secert_number. keep trying!")
    elif (user_input < secert_number):
        print("Your number is lesser than the secert_number. keep trying!")
    else:
        print(f "Congratulations! you guessed the secert number {secert_number} in {i+1} attempts")
        break
```

## 6. using while loop to execute the code until you find the secert number and handle the exception using try and except block



```python
import random
print(" WELCOME TO THE NUMBER GUESSING GAME!! ")
print(" Guess the number between 1 to 100. ")
print("*************************************************")

secert_number = random.randint(1,100)
count = 0
while True:
    try:
        user_input = int(input("Enter the number:"))
        count += 1
        if(user_input >100 or user_input < 1):
            print("Your number is invalid enter the number between 1 - 100.")
        elif (user_input > secert_number):
            print("Your number is greater than the secert_number. keep trying!")
        elif (user_input < secert_number):
            print("Your number is lesser than the secert_number. keep trying!")
        else:
            print(f"Congratulations! you guessed the secert number {secert_number} in {count} attempts")
            break
    except ValueError:
        print("enter the numbers between 1 to 100.") 
```
