### 2.  Set and it's inbuilt functions
1. 1D
2. indexing and silicing is not applicable
3. mutable
4. allows all kinda datatypes

```python
colors = {"Blue","Yellow","orange","red"}
colors
```

```python
colors.add("Black")
colors
```

```python
colors.copy()
colors
```

```python
neon_colors ={"fluorescent","Hot Pink","orange"}
neon_colors
```

```python
colors.difference(neon_colors) # doesnt get updated in original set
```

```python
colors
```

```python
update_colors=colors.difference_update(neon_colors) # gets updated in original set
colors
```

```python
colors.discard("red")
colors
```

```python
neon_colors.add("Blue")
neon_colors
```

```python
colors.intersection(neon_colors) # doesnt get updated the common value in original set 
```

```python
colors
```

```python
colors.intersection_update(neon_colors) #gets updated the common value in original set and only the common value will be present int the set
colors
```

```python
colors.isdisjoint(neon_colors) #Return True if two sets have a null intersection.
```

```python
colors.issubset(neon_colors)
```
