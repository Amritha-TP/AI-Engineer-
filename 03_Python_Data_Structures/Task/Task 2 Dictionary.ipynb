## Task 2: Create 3 DataFrames using Dictionary

```python
import pandas as pd
```

### 1. List in Dictionary
 allows duplicate
- allows mixed types in the list
- 1 Dimensional
- mutable --> the values can be changed by accessing the index value
- can access data via indexing
  * forward accessing starts with index 0
  * backward accessing starts with index -1

```python
student_grade_details = {"Name":["Viyaan","Amritha","Bala","Rio"],
                         "class":[10,10,10,10],
                         "City":["chennai","Theni","Coimbatore","Theni"],
                         "Grade":["A","A","C","B"],
                         "Result":["Pass","Pass","Fail","Pass"]
                        }
Student_table=pd.DataFrame(student_grade_details)
```

### 2. Tuple in Dictionary
1. 1d
2. allow duplicate
3. allow all datatype
4. indexing and slicing possible
5. immutable - values cannot be changed once assigned

```python
restaurant_menu = {"Item_no":(1,2,3,1),
                   "Food_Items": ("Biryani", "Pizza", "Burger","Biryani"),
                   "Price":("$30","$9.99","$8.99","")
                  }

menu_table = pd.DataFrame(restaurant_menu)
menu_table
```

### 3. Set in Dictionary

* 1D
* indexing and silicing is not applicable
* mutable
* allows all kinda datatypes
* set cannot be directly converted into rows and columns using DataFrame.
   - set has to be convereted to list or tuple before converting it to table.



```python
icecream_shop = {
                 "Icecream_ flavors":{"strawberry","chocolate","Butter scotch"},
                 "Quantity":{"Small","Medium","Large"},
                 "Topping":{"Choco-chip","dry-fruits","maple-syrup"}
                }
                


pd.DataFrame(list(icecream_shop)) # displays the key mentioned
```

```python
icecream_shop = {
                 "Icecream_ flavors":{"strawberry","chocolate","Butter scotch"},
                 "Quantity":{"Small","Medium","Large"},
                 "Topping":{"Choco-chip","dry-fruits","maple-syrup"}
                }
                


pd.DataFrame(list(icecream_shop["Icecream_ flavors"])) # displays the value of the key mentioned
```

```python
icecream_shop = {
                 "Icecream_ flavors":{"strawberry","chocolate","Butter scotch"},
                 "Quantity":{"Small","Medium","Large"},
                 "Topping":{"Choco-chip","dry-fruits","maple-syrup"}
                }
                


# converting entire set dict to list dict so pandas can convert to row and columns

icecreamshop_menu = {
                     key : list(value) for key, value in icecream_shop.items()
                    }

pd.DataFrame(icecreamshop_menu)
```
