## Dictionary - {}
1. key : value pair
2. indexing and silicing is not possible
3. mutable
4. it allows duplicate values and the key must be unique
5. allows all datatypes
6. 2D

```python
empty_cart_dict={} #empty dictionary
```

```python
type(empty_cart_dict)
```

```python
empty_cart_dict
```

```python
items_tobuy_dict ={"Math":100, "science":99, "Computer_Science":99}
```

```python
items_tobuy_dict.keys()
```

```python
items_tobuy_dict.values()
```
