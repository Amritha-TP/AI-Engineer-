## Object Detection using YOLO model

```python
from ultralytics import YOLO #loads and runs YOLOv8 models
import cv2 #OpenCV for video capture and drawing boxes

model = YOLO("yolov8n.pt") #Loads the YOLOv8‑Nano model 

video = cv2.VideoCapture(0) # 0 means default webcam and 1 means external cams like cctv

while True: #Runs continuously until you manually stop it
    
    success, frame = video.read() #frame = one image from webcam
    if not success:
        break
        
    '''
    YOLO processes the frame 
    Returns detected objects, bounding boxes, class IDs, confidence scores
    '''
    results = model(frame)

    for result in results:
        boxes = result.boxes # boxes contains coord, classID, confidence
        for box in boxes:
            x1, y1, x2, y2 = map(int, box.xyxy[0]) #box.xyxy has coord like x1,y1 etc
            confidence = float(box.conf[0])
            class_id = int(box.cls[0])
            class_name = model.names[class_id]
            cv2.rectangle(frame, (x1, y1), (x2, y2), (255, 0, 0), 2)
            cv2.putText(frame, f"{class_name} {confidence:.2f}", (x1, y1 - 10),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.7, (255, 255, 255), 2)
    cv2.imshow("YOLO Webcam Detection", frame) #opens webcam
    key = cv2.waitKey(1) #1000FPS waitKey(1) keeps the loop responsive and fast.
    if key == 81 or key == 113:
        break

video.release()
cv2.destroyAllWindows()
print("Video Stopped")

'''
Note:

waitKey() is the heartbeat of OpenCV video display.

It:

updates the window

listens for key presses

controls timing

keeps the GUI alive
'''
```
