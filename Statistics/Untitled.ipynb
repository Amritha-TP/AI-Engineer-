```python
import numpy as np
from scipy import stats

# ── Problem parameters ──────────────────────────────────────────
mu0   = 65      # historical population mean (H0)
xbar  = 68      # desired sample mean
s     = 10      # desired sample std
n     = 50      # sample size
alpha = 0.05    # significance level

# ── Step 1: Generate a sample with EXACT mean=68 and std=10 ─────
np.random.seed(42)
raw = np.random.normal(loc=xbar, scale=s, size=n)

# Standardize → mean=0, std=1
raw = (raw - raw.mean()) / raw.std(ddof=1)

# Re-scale  → mean=68, std=10  (matches problem statement exactly)
sample = raw * s + xbar

print("── Sample verification ──────────────────────")
print(f"  Sample mean : {sample.mean():.4f}")      # exactly 68.0
print(f"  Sample std  : {sample.std(ddof=1):.4f}") # exactly 10.0

# ── Step 2: One-sample t-test (one-tailed, right tail) ──────────
# H0: mu <= 65  (course doesn't help)
# H1: mu >  65  (course improves scores)

t_stat, p_value = stats.ttest_1samp(
    a         = sample,
    popmean   = mu0,
    alternative = 'greater'   # one-tailed, right tail
)

t_crit = stats.t.ppf(1 - alpha, df=n-1)   # critical value

print("\n── Hypothesis Test Results ──────────────────")
print(f"  t-statistic    : {t_stat:.4f}")
print(f"  t-critical     : {t_crit:.4f}")
print(f"  p-value        : {p_value:.4f}")
print(f"  alpha          : {alpha}")

print("\n── Decision ─────────────────────────────────")
if p_value < alpha:
    print("  Reject H0 ✓")
    print("  Significant evidence that the new course improves performance.")
else:
    print("  Fail to reject H0 ✗")
    print("  Insufficient evidence of improvement.")
```
