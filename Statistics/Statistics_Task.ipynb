### HYPOTHESIS TESTING

Hypothesis testing is a statistical method used to make decisions using sample data with the help of our intitution.

Null hypothesis (H0): nothing changed, no effect, no difference

Alternative hypothesis (Ha): something changed, there is an effect or difference

We collect data and calculate a p‑value, which tells us how likely our sample is if H0 were true.

We compare the p‑value with a threshold called alpha or level of significance, usually 5% or 0.05:

If p‑value < level of significance → reject H0 (evidence supports Ha)

If p‑value >= level of significance → fail to reject H0 (not enough evidence)

We never say “accept H₀” because lack of evidence is not proof.

### STATISTICAL TESTING

The test is performed to prove the alternate hypothesis and reject the null hypothesis by knowing the pvalue.

One-Sample (Dependent): Used when comparing a single group's mean or proportion.

T-test (Mean): Divided into 1-tailed or 2-tailed tests.

Proportion test (%): Also divided into 1-tailed or 2-tailed tests.

Two-Sample (Independent): Used when comparing two completely separate, independent groups to see if their means or proportions differ significantly.

T-test (Mean): Can be 1-tailed or 2-tailed.

Proportion test (%): Can be 1-tailed or 2-tailed.

Paired-t-test (Dependent): Used when comparing two dependent or related groups (for example, measuring the same group of subjects "before and after" a specific treatment). use ztest because you are compare=ing two different propotion.

ANOVA-test (Independent): Used when you want to compare the means of three or more independent groups simultaneously.

Chi-squared Test: Typically used for categorical data to determine relationships between two categories.

#### Q1: The historical average score in a certification exam is 65. A training institute introduces a new course. A random sample of 50 students trained under the new course has a mean score of 68 and a standard deviation of 10. At the 5% level of significance, test whether the new course improves performance.

-  mean of pop is 65.  mean of sample is 68.  SD is 10. LOS is 0.05
-  H0 >= 65
-  Ha <65
-  T-test because we are using mean of scores not %

```python
from scipy import stats
import numpy as np

#Same seed → same random numbers. Different seed → different random numbers.
np.random.seed(42)
raw = np.random.normal(loc=68, scale=10, size=50) # give random sample value but not exact mean and sd.

# Force exact mean and std
raw = (raw - raw.mean()) / raw.std(ddof=1)  # standardize -zero score normalization wher mean =0 std =1
raw = raw * 10 + 68                          # rescale to mean=68, std=10
students_obs_50 = raw.astype("int")

statistics, pvalue = stats.ttest_1samp(a=students_obs_50, popmean=65, alternative='greater')

print(f"p-value     : {pvalue:}")    

if pvalue < 0.05:
    print("Reject H0, Significant evidence that the new course improves performance.")
else:
    print("Fail to reject H0, Insufficient evidence of improvement.")
```

#### Q2: A company claims that the average time taken to resolve customer tickets is 30 minutes. A random sample of 40 tickets shows a mean resolution time of 33 minutes with a standard deviation of 8 minutes. At the 5% level of significance, test whether the average resolution time has increased.

```python
from scipy import stats
import numpy as np

p_mean_2 = 30
s_mean_2 = 33
n_2 = 40
sd_2 = 8
los_2 = 0.05

#one sample t-test mean, onetailed(right)

np.random.seed(43)

raw_2 = np.random.normal(loc=s_mean_2,scale=sd_2,size=n_2)

raw_2 = (raw_2-raw_2.mean()) / raw_2.std(ddof=1)

raw_2 = raw_2 * sd_2 + s_mean_2
print("raw mean:",raw_2.mean())

cust_ticket_time_obs = raw_2.astype("int")

statistics,pvalue=stats.ttest_1samp(a=cust_ticket_time_obs, popmean=p_mean_2, alternative='greater')

print("p-value:",pvalue)

if pvalue < 0.05:
    print("Reject H0, Significant evidence that the average resolution time has increased.")
else:
    print("Fail to reject H0, Insufficient evidence that the average resolution time has increased.")
```

#### Q5: Two teaching methods are used to train employees. A sample of 30 employees trained using Method A has a mean score of 75 with a standard deviation of 12. A sample of 28 employees trained using Method B has a mean score of 70 with a standard deviation of 10. At the 5% level of significance, test whether there is a difference in average scores between the two methods.

```python
from scipy import stats

#h0 -> mean A = mean B
#h1 -> mean A != mean B # two_sample two-tail test
#This is a two‑tailed test because you're checking for any difference, in either direction. not which method is best.
t_stats_tm,p_value_tm = stats.ttest_ind_from_stats(
    mean1=75,std1=12,nobs1=30,
    mean2=70,std2=10,nobs2=28,
    equal_var=True,  # Assumes equal population variance; set to False if variances differ significantly
    alternative='two-sided')

print(f"t_statistic: {t_stats_tm}\npvalue: {p_value_tm}")

if p_value_tm <0.05:
    print("Reject H0, Significant evidence that there is a difference between the two teaching methods.")
else: 
    print("Fail to reject H0, Insufficient evidence to claim a difference between the two methods.")
```

```python
from scipy import stats

#h0 -> mean A = mean B
#h1 -> mean A != mean B # two_sample two-tail test
#This is a two‑tailed test because you're checking for any difference, in either direction. not which method is best.
t_stats_del,p_value_del = stats.ttest_ind_from_stats(
    mean1=48,std1=6,nobs1=35,
    mean2=52,std2=7,nobs2=40,
    alternative='two-sided')

print(f"t_statistic: {t_stats_del}\npvalue: {p_value_del}")

if p_value_del <0.05:
    print("Reject H0, Significant evidence that there is a difference in the delivery time between warehouses A and B.")
else: 
    print("Fail to reject H0, Insufficient evidence to claim a difference in the delivery time between  warehouses A and B.")
```

#### Q7: Two versions of a website are tested. Out of 500 users for Version A, 320 users click the signup button. Out of 480 users for Version B, 290 users click the signup button. At the 5% level of significance, test whether the conversion rates differ.

```python
import statsmodels.api as sm

#H0 -> p_A = p_B
#H1 -> p_A != p_B so two tailed, two sample propotion ztest

count1=[320,290]
nobs1=[500,480]

z_stat_web,pvalue_web = sm.stats.proportions_ztest(
    count=count1,
    nobs=nobs1,
    alternative='two-sided'
)

print(f"z_stat: {z_stat_web}\npvalue: {pvalue_web}")

if pvalue_web <0.05:
    print("Reject H0, Significant evidence that there is a conversion rate difference between websites A and B.")
else: 
    print("Fail to reject H0, Insufficient evidence to claim that there is a conversion rate difference between websites A and B.")

```

#### Q9: A company wants to compare the average productivity of employees trained using three different training programs. Program A has 25 employees with a mean productivity score of 80 and a standard deviation of 8. Program B has 30 employees with a mean of 75 and a standard deviation of 7. Program C has 28 employees with a mean of 82 and a standard deviation of 6. At the 5% level of significance, test whether there is a difference in mean productivity across the programs.

```python
import numpy as np
import statsmodels.stats.oneway as ow

# one way annova because the 3 tests are independent
# h0 = all means are equal
#h1 = one group mean is different
"""
The F-statistic in ANOVA (Analysis of Variance) is a ratio comparing the amount of variability between different groups to the amount of variability within those groups. 
It tells you if the differences between your group averages are statistically meaningful or likely just random chance
"""

n_A, mean_A, sd_A = 25, 80, 8
n_B, mean_B, sd_B = 30, 75, 7
n_C, mean_C, sd_C = 28, 82, 6

# anova_generic requires numpy arrays, not lists. anova generic is used when u known the mean sd, sample size
means     = np.array([mean_A, mean_B, mean_C])
variances = np.array([sd_A**2, sd_B**2, sd_C**2])   # variances, not std deviations (varience = 2 to the power of sd)
nobs      = np.array([n_A, n_B, n_C])

# use_var='equal'  classic one-way ANOVA (equal variances assumed)
res = ow.anova_generic(means=means, variances=variances, nobs=nobs, use_var='equal')

f_statistic = res.statistic
pvalue_9    = res.pvalue

print(f"F-statistic: {f_statistic}")
print(f"p-value    : {pvalue_9}")

if pvalue_9 < 0.05:
    print("Reject H0: Significant evidence that there is a difference in mean productivity across the programs.")
else:
    print("Fail to reject H0: Insufficient evidence to claim a difference across the programs.")
```

#### Q10:     10. A company wants to test whether gender and product preference are independent. The survey results are as follows:
    • Male: 60 prefer Product A, 40 prefer Product B
    • Female: 50 prefer Product A, 50 prefer Product B
    At the 5% level of significance, test for independence.

```python
#two categorical data so chi square test
#stats.chisquare() is only for 1‑D goodness‑of‑fit tests, not for 2×2 or larger contingency tables.
```

```python
import scipy.stats as stats
import numpy as np
observed = np.array([[60,40],
                  [50,50]])
result = stats.chi2_contingency(observed=observed,correction=False) # correction=False - no yate's adjustment
print (f"statistic: {result.statistic}\npvalue: {result.pvalue}")

if result.pvalue>0.05:
    print("Fail to reject H0, as there is sufficient evidence that the categories are independent.")
else:
    print("Reject H0, as there is sufficient evidence that the categories are dependent.")
```

#### Q12:     12. A dice manufacturer claims that a die is fair. The die is rolled 120 times, and the observed frequencies are:
    • Face 1: 18
    • Face 2: 22
    • Face 3: 20
    • Face 4: 25
    • Face 5: 17
    • Face 6: 18
At the 5% level of significance, test whether the die is fair.


```python
# chisquare test
#h0=dies test is fair
#h1=dies test is unfair

import scipy.stats as stats
import numpy as np
observed = np.array([18,22,20,25,17,18])
expected = np.array([120 / 6] * 6)
result = stats.chisquare(f_obs=observed,f_exp=expected) # correction=False - no yate's adjustment
print (f"statistic: {result.statistic}\npvalue: {result.pvalue}")

if result.pvalue>0.05:
    print("Fail to reject H0, dies test is fair.")
else:
    print("Reject H0, dies test is unfair")
```
