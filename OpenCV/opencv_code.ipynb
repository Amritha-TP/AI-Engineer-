## Single/Multiple Object Detection
 Detect a single object in an image 
 \Detect multiple objects in an image and visualize bounding boxes

```python
import cv2
cv2.__version__
```

```python
face_cascade = cv2.CascadeClassifier("haarcascade_frontalface_alt.xml")

image=cv2.imread("group of people.jpg")
grey_image = cv2.cvtColor(image,cv2.COLOR_BGR2GRAY)
face_detector = face_cascade.detectMultiScale(grey_image)
face_detector

for x,y,w,h in face_detector:
    cv2.rectangle(grey_image,(x,y),(x+w,y+h),(0,0,0),6)

cv2.imshow("Image show window",grey_image)
cv2.waitKey()
```

## Webcam / CCTV Detection
Real-time detection using webcam feed

```python
import cv2
face_cascade = cv2.CascadeClassifier("haarcascade_frontalface_alt.xml")

video = cv2.VideoCapture(0)

while(True):
    sucess,frames = video.read()

    face_detector = face_cascade.detectMultiScale(frames)
    for x,y,w,h in face_detector:
        cv2.rectangle(frames,(x,y),(x+w,y+h),(0,0,0),2)
    cv2.imshow("Video show window",frames)
    key = cv2.waitKey(1)
    if key == 81 or key==113:
        break
video.release()
cv2.destroyAllWindows()
```
