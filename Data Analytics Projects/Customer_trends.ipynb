```python
customer_data['Review Rating']=customer_data.groupby('Category')['Review Rating'].transform(lambda x : x.fillna(x.mean()))
```

```python
customer_data.isna().sum()
```

```python
customer_data.columns = customer_data.columns.str.lower()
customer_data.columns = customer_data.columns.str.replace(" ","_")
customer_data.rename(columns={'purchase_amount_(usd)':'purchase_amount'},inplace=True)
customer_data.columns
```

```python
labels = ['Young Adult','Adult','Middle Aged','Senior']
customer_data['age_group'] = pd.qcut(customer_data['age'],q=4,labels=labels)
```

```python

frequency_mapping = {'Fortnightly':14,
                     'Weekly':7,
                     'Annually':365,
                     'Quarterly':90,
                     'Bi-Weekly':14,
                     'Monthly':30,
                     'Every 3 Months':90}

customer_data['purchase_frequency_days'] = customer_data['frequency_of_purchases'].map(frequency_mapping)
```
